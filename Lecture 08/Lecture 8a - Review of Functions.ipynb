Example: Calculate

$V=P\left(1+{\frac {r}{n}}\right)^{nt}$




````python
# We are going to define a function "fn_compute_value"
# You can choose any name
# Using prefixes like "fn_" can help you remember this is a "function" object
# What are the parameters?

def function_v(P, r,n, t):
    v = P*(1+(r/n))**(n*t)
    return v
````

````python
function_v(P=100, r = 0.01, n = 20, t=10)
````

````python
# You can know compute the formula with different values

V1 = function_v(P = 1000, r = 0.01, n = 20, t=10)
V2 = function_v(P = 10, r = 0.01, n = 20, t=10)

print(V1)
print(V2)
````

## <span style="color:darkblue"> III. Lambda Functions </span>

<font size = "5">

"Lambda Functions" are defined in one line:

```python
my_function = lambda parameters: expression
```

<font size = "5">

Example: Calculate $x + y + z$

````python
# (a) Define function
fn_sum = lambda x,y,z: x + y + z

# (b) Run function
fn_sum(1,2,3)
# lambda creates a small, unnamed function. In this case,the inputs are x,y, and z and return their sum. 
so for the fnsum, use the corresponding values 1,2,3 for the variables. 

#lambda is a compact one line function that is useful for short operations. 
#custom def functions - contain multiple statemets such as variables, loops, and if statements. Better for complex or reusable logic.
````

<font size = "5"> Example: Calculate

$V=P\left(1+{\frac {r}{n}}\right)^{nt}$


````python
fn_v = lambda P, r, n, t: P*(1+(r/n))**(n*t)
````

````python
V1 = fn_v(P = 1000, r = 0.01, n = 20, t=10)
V2 = fn_v(P = 10, r = 0.01, n = 20, t=10)

print(V1)
print(V2)
````
