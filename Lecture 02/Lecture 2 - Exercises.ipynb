    - All the histograms should be overlapping

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data_wdi_urban = pd.read_csv(r"C:\Users\Owner\Documents\GitHub\datasci530fall2026\Lecture 02\data_raw\wdi_urban.csv")

list_unique_years = pd.unique(data_wdi_urban["year"])
print(list_unique_years)

for year in list_unique_years:
    subset = data_wdi_urban.query("year == @year")
    plt.hist(subset["prop_urbanpopulation"], alpha=0.5, label=str(year))

plt.legend(title="Year")
plt.xlabel("Proportion of Urban Population")
plt.ylabel("Frequency")
plt.title("Distribution of Urban Population Share, by Year")
plt.show()

From the graph, we can see that urban population has generally increased across the distribution of countries in 2020 with more spread across the higher end. In the 1980 histogram, urban populations are concerntrated more toward the lower to mddle range.  

````

(k) Composite operations on arrays

<font size = "3">

- Define $r = 0.04$
- Define two arrays $A$ and $T$ using "np.array()", assigning the following values

$ a = \begin{pmatrix} a_0 \\ a_1 \\ a_2 \end{pmatrix} =  \begin{pmatrix} 20 \\ 30 \\ 15 \end{pmatrix},
\qquad  t = \begin{pmatrix} t_0 \\ t_1 \\ t_2 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \\ 2 \end{pmatrix} $

- Use element-by-element vector operations to compute a new array $M$ of length 3:

$ M = \begin{pmatrix} a_1 / (1+r)^{t_0}  \\ a_2 / (1+r)^{t_1} \\ a_3 / (1+r)^{t_2} \end{pmatrix} $

- Compute the sum of the values of $M$ using "M.sum()"

Note: I am using subscripts to denote the individual elements of $a$ and $t$.
- In the code just enter the values of $a$ and $t$.
- To get full points you must do all of the above tasks



````python
import numpy as np

r = 0.04

A = np.array([20, 30, 15])
T = np.array([0, 1, 2])

M = A / (1 + r)**T

total = M.sum()
print(total)





````
