# <span style="color:darkblue"> II. Generate simulated Data </span>

<font size = "5">

Create an empty dataset

````python
dataset = pd.DataFrame()
````

<font size = "5">

Create two random variables of size ($n = 50$)

````python
n = 50
np.random.seed(42)
#this adds in two datasets
dataset["x"] = np.random.normal(loc = 0,scale = 1, size = n)
dataset["e"] = np.random.normal(loc = 0,scale = 1, size = n)
````

<font size = "5">

Create data from the linear model

$ y = b_0 + b_1 x + e, \qquad b_0 = 1, b_1 = 2.$

````python
# The number b0 is known as the "intercept"
# The number b1 is known as the "slope"

b0 = 1
b1 = 2

# We can compute formulas directly over dataset columns
dataset["y"] = b0 + b1 * dataset["x"] + dataset["e"]
````

````python
dataset['y']
````

<font size = "5">

Compute the theoretically best fit line

$ p = b_0 + b_1 x$

````python
dataset["p"] = b0 + b1*dataset["x"]
````

<font size = "5">

Plot the data

````python
plt.scatter(x = dataset["x"], y = dataset["y"])
plt.plot(dataset["x"],dataset["p"], color = 'green')

plt.xlabel("X Variable")
plt.ylabel("Y Variable")
plt.legend(labels = ["Data points", "Best fit line"])
plt.show()
````

# <span style="color:darkblue"> II. Estimate the best fit line</span>


<font size = "5">

We have data on $(y,x)$ but we don't know $(b_0,b_1, e)$

<font size = "5">

Let's fit an OLS model

- It's a statistical approach to get $(b_0,b_1)$
- No need to know how it works but why we want it

````python
#------------------------------------------------------------------------------#
# We use the subfunction "ols()" in the library "smf"
#---- (i) The first argument is a string called "formula" with the format 
#-------- "outcome ~ indepdent_vars"
#----(ii) the second argument is the dataset
# The second line fits the model with standard errors "cov". In this case we 
# use "robust" standard errors (HC1)
#-------------------------------------------------------------------------------#

model = smf.ols(formula = 'y ~  x', data = dataset)
#first state what kind of model we want - ths case stats model formula (ordinary least squares for line procedures)
#we will also input a formula and a dataset
#first step is configuration, not calculating anything
#this formula is an abbreviated format and the intent of this function is to set up a model showing how x relates to y in my dataset; y~x means use x to predict or explain y
#data =dataset tells python where to find the x and y columns (in the dataset)
#Smf.ols(...)- creates an ordinary least squares linear regression model - the kind that finds the straight line that best fits the dataset
#smf is the nickname or alias for the statsmodels.formula.api module; often comes from an import like statsmodels.formula.api as smf
# results = model.fit()
results = model.fit(cov = "HC1")
#this is ok to use all of the time 
#???////////
#"HC1" is a method for calculating robust standard errors, which can help when the model’s errors vary in size across observations.
````

<font size = "5">

Compute the estimated best fit line

````python
# We will use ".params" to get the attribute "parameters from the results"
#this is the next step after "fitting" the regression model using results = model.fit()
#results.params gives you the numbers that define the best-fitting line
#REsults.params collects the model's estimated values- usually the **intercept** and the **slope** for x
#b_LISt - savws those values under the name b_list
#for a model y~x, the line is roughly predicted y=intercept +slope times x; slope tells you how much the predicted y changes when x increases by one 
b_list = results.params
print(b_list)

# We can then compute the "estimated" best fit lines
# by extracting the intercept and slop from "b_list"

dataset["p_estimated"] = b_list[0] + b_list[1]  * dataset["x"]

# Note: The estimators for "b0" and "b1" are close to 
# the values we used to generate the data
````

<font size = "5">

Plot the best fit line

````python
# Use scatter twice, with different "y" inputs
# THe "legend" command creates a box on with the color labels

plt.scatter(x = dataset["x"],y = dataset["y"])
plt.plot(dataset["x"],dataset["p_estimated"], color = 'green')

plt.legend(labels = ["Data points","Estimated Predicted Model"])
plt.show()
````
